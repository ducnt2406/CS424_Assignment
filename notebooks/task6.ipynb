```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("../data/Processed_UICLibraryData.csv")
df["occupancy_index"] = df.groupby("floor")["occupancy_count"].transform(lambda x: x / x.iloc[0] * 100)
df

pivot = df.pivot(index="date", columns="floor", values="occupancy_index")
```

```python
pivot.plot(kind="line", marker="o", subplots=True, layout=(1, 2), figsize=(10, 5), legend=False, sharey=True, title=["First floor", "Second floor"])[0,1].tick_params(labelleft=True)
plt.ylabel("occupancy_count index (first day = 100)")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(data=df, x="date", y="occupancy_index", hue="floor", marker="o")

plt.ylabel("Occupancy index (first day = 100)")
plt.show()
```
