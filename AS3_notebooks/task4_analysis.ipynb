```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/AS3_evaluation_result.csv")
```

```python
res = df.groupby("condition").agg(
    mean_accuracy=("correctness", "mean"),
    mean_time=("time_seconds", "mean")
)
print(res)
```

```python
taskres = df.groupby(["condition", "task"]).agg(
    mean_accuracy=("correctness", "mean"),
    mean_time=("time_seconds", "mean")
)
print(taskres)
```

```python
acc = df.groupby("condition")["correctness"].mean()

acc.plot(kind="bar")
plt.ylabel("Mean accuracy")
plt.title("Accuracy: Static vs Interactive")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```

```python
time = df.groupby("condition")["time_seconds"].mean()

time.plot(kind="bar")
plt.ylabel("Mean completion time (seconds)")
plt.title("Completion Time: Static vs Interactive")
plt.tight_layout()
plt.show()
```

```python
summary = df.groupby(
    ["participant_id", "condition"]
).agg(
    accuracy=("correctness", "mean"),
    mean_time=("time_seconds", "mean")
)

print(summary)
```
